- Let us take an analogy of **Airport** where there should be **only one central control tower** which ensures everything happens in orderly manner.
- Having multiple control towers at one airport would cause chaotic and conflicting instructions for pilot.

### Bad Example : No Singleton

````python
class ControlTower:
    def __init__(self):
        print("Initializing Control Tower ✈️!!!")
````

````python
tower1 = ControlTower()
tower2 = ControlTower()
````

- Here , each time the class `ControlTower()` is invoked, new objects , i.e new instances of the class is being created, which can be verified by the print statement as well.

- We have `Initializing Control Tower ✈️!!!` printed each time the class `ControlTower()` is invoked ; twice in this example.

- By definition, we know `__init__`is a special method that gets **automatically called each time an object is created.**

````python
print(tower1 is tower2) # False means they are not the same instance of the class.
````

````python
print(f"Tower1: {tower1}")
print(f"Tower2: {tower2}")
````

--------------------

- Our Aim : An airport should have only one control tower.

- Lets say, in our code we mistakenly created another instance of `ControlTower` as `tower2`.

- We want our code such that no matter how many times the class `ControlTower` is invoked, it should only have a single instance.

- `tower1` & `tower2` should be the same instance which was not the case above.

---------------

#### Before Getting into solutions, try to understand the concept.

When we want to create an instance from a class , we write :

```
obj = MyClass()
```
The following things are happens internally in python:
```
obj = MyClass.__new__(MyClass)   # step1 : Create the object
MyClass.__init__(obj)            # step2 : Initialize the object
```
💡Do not get overwhelmed by the syntax, just know that:
1. `__new__` method is used to : Allocate memory , create and return the raw Object (or Instance).
2. `__init__` method is used to : Initialize the object **after it is created.**
>NOTE:
>
> -  `__new__` method is called before `__init__` method.
> - Also remember, while `__init__` is often called the **'Constructor'** casually , in strict sense; `__new__` is the **true Constructor** because it constructs the object , `__init__` is the **Initializer**.

-----------------

### Solution : Singleton Pattern

````python
class ControlTower:
    instance = None
    def __new__(cls):
        if cls.instance is None:
            cls.instance = super().__new__(cls)
            print("Initializing Control Tower ✈️!!!")
        return cls.instance
        
````

````python
tower_x = ControlTower()
tower_y= ControlTower()


# We will have 'Initializing Control Tower ✈️!!!' printed only once.
# Eventhough the class ControlTower() was invoked twice in this code.
````

````python
print(tower_x is tower_y)
````

````python
print(f"First  Tower: {tower_x}")
print(f"Second Tower: {tower_y}")
````
