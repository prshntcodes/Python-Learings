## Factory Pattern Core Idea:
- Create Objects through a factory , **not directly** in the client code

### Bad Example

```python
from abc import ABC, abstractmethod
# Interface aka Base class
class Molecule(ABC):
    @abstractmethod
    def analyze(self):
        pass

# Concrete Classes
class Protein(Molecule):
    def analyze(self):
        print("Analyzing Protein 🍗🍗 ")

class DNA(Molecule):
    def analyze(self):
        print("Analyzing DNA 🧬🧬 ")

# Client Code
# Problem: Object-creation logic is hardcoded in the client.
# This creates tight coupling and makes the code harder to extend.

class Analysis: # Responsible for object creation
    def perform_analysis(self, mol_type):
        if mol_type == "dna":
            molecule = DNA()
        elif mol_type == "protein":
            molecule = Protein()
        else:
            print ("❌ Invalid Molecule, Cannot Perform Analysis ❌")
            return None

        molecule.analyze()
        #return molecule

# Use case:
lab_work = Analysis()
lab_work.perform_analysis("protein")
```

------------------
#### Q. How to achieve Factory Pattern ?
- Define a common interface or base class.
- Create concrete classes that implement that interface.
- Write a factory method or factory class that decides which object to return.
- Let the client code use the factory instead of creating objects directly.
---------------

```python
from abc import ABC, abstractmethod
# Step 1 : Common Interface 
class Molecule(ABC):
    @abstractmethod
    def analyze(self):
        pass
#---------------------------------------------------------
# Step 2 :Concrete Classes
#---------------------------------------------------------
class Protein(Molecule):
    def analyze(self):
        print("Analyzing Protein 🍗🍗 ")

class DNA(Molecule):
    def analyze(self):
        print("Analyzing DNA 🧬🧬 ")

#---------------------------------------------------------
# Step 3 : Factory
#---------------------------------------------------------
class MoleculeFactory:
    @staticmethod
    def create(mol_type):
        if mol_type == "dna":
            return DNA()
        elif mol_type == "protein":
            return Protein()
        else:
            print (f"❌Invalid molecule type: {mol_type}❌") 
            return None

#---------------------------------------------------------
# Step 4: Client code: 
# The client asks MoleculeFactory to create the appropriate molecule 
# rather than directly creating DNA() or Protein().
#---------------------------------------------------------
class Analysis: 
    def perform_analysis(self, mol_type):
        molecule = MoleculeFactory.create(mol_type)
        if molecule is None:
            print("Cannot Perform Analysis on Invalid Molecule")
            return
        
        molecule.analyze()


analysis_lab = Analysis()

analysis_lab.perform_analysis("dna")
analysis_lab.perform_analysis("protein")
analysis_lab.perform_analysis("sugar")
```
