### Reading data from a text file
To read a file, we first need to open it. We can use the built-in **open()** function to open a file in Python. 
The open() function takes 2 arguments: 
1. the name of the file or path of the file and 
2. the mode in which we want to open the file. 
   * 'r' for reading, 
   * 'w' for writing, 
   * 'a' for appending, and 
   * 'b' for binary mode.

* In **read** mode : we can only read and print the data of the file but cannot write in the file.
  
* In **write** mode : it is prefered if we want to create a **new file** and write some data or information in the file. If we open an existing file in write mode,the content in the file is automatically erased and replaced with the new data we write.

  
* In **append** mode : if we want to add or write something in an existing file without overwriting the previous content of the file . It ddds new data to the end of the file without deleting existing content. 

### Steps to consider : Open the file -> Read it -> Close the file
* Do not forget to close the file after reading it. It is a good practice to close the file after you are done with it.
* Not closing the file is a common **beginner's mistake** and we should look to avoid it. It can lead to memory leaks and resource related issues.

```python
file = open("dummy_data.txt", "r") # file is opened in read mode, replace "dummy_data.txt" with your file_name/ filepath.
data = file.read() # file is read and stored in the variable 'data'
file.close() # file is closed after reading it
print(data) 
```

```python
file = open("dummy_data.txt", "r") 
file.close() 
data = file.read()  #-> it will cause error as we cannot read a closed file.
print(data) 
```

```python
file = open("dummy_data.txt","r")
for line in file:
    print(line)   # "\n" new line characeter is present after each line by default.
# By default it is;  print(line, end="\n")
file.close()
```

### Concept
* the **.read( )** function reads the entire file as a **single string**.
* the for loop reads the file line by line

```python
file = open("dummy_data.txt","r")
i=1
for line in file:
    print(i,line, end= "")
    i+=1
file.close()
```

### No need to manually close the file each time we open it. 
* There is a better way to do it in python, using **with** keyword. It closes the file automatically.

* The with statement in Python simplifies resource management by automatically handling setup and cleanup, ensuring files or connections close safely even if errors occur.

```python
with open("dummy_data.txt","r") as file:
    for line in file:
        print(line.strip())
```

## write mode : it will write the new data but it will automatically overwrite the previous data/information of the file.
*  This is it's drawback, any existing content in the file is **automatically erased and replaced** with the new data you write

```python
with open("dummy_data.txt","w") as file:
    file.write("--This is a new line I have added in Write mode.--")
```

```python
# look at what the file holds now
with open("dummy_data.txt","r") as f:
    print(f.read())
```

### To address this problem, we have append mode.

```python
with open("dummy_data.txt","a") as f:
    f.write("\n A new line is appended, without erasing previous content of the file.")
```

```python
with open("dummy_data.txt","r") as f:
    print(f.read())
```
